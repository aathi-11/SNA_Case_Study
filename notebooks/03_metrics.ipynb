# 03. Network Metrics & Centrality Analysis

This notebook analyzes network centralities and parametric structural metrics across all **256 team-match passing graphs**.

```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"

team_metrics = pd.read_parquet(PROCESSED_DIR / "team_network_metrics.parquet")
player_metrics = pd.read_parquet(PROCESSED_DIR / "player_network_metrics.parquet")

print(f"Loaded {len(team_metrics)} team network metric summaries.")
print(f"Loaded {len(player_metrics)} player centrality metric records.")
```

## 1. Team-Level Network Metrics Overview

```python
metric_cols = [
    "density", "reciprocity", "degree_centralization",
    "betweenness_centralization", "weighted_clustering",
    "global_efficiency", "playmaker_share"
]

print("--- Summary Statistics of Structural Metrics ---")
display(team_metrics[metric_cols].describe())
```

## 2. Structural Metrics by Match Outcome (Win vs Loss)

```python
win_loss_df = team_metrics[team_metrics["result"].isin(["Win", "Loss"])]
grouped = win_loss_df.groupby("result")[metric_cols].mean().reset_index()
print("--- Mean Metrics: Winners vs Losers ---")
display(grouped)
```

## 3. Top Playmakers by Betweenness Centrality

```python
top_playmakers = player_metrics.sort_values("betweenness_centrality", ascending=False).head(15)
print("--- Top 15 Highest Betweenness Centrality Players (Single Match) ---")
display(top_playmakers[["match_id", "team_name", "player_name", "betweenness_centrality", "eigenvector_centrality", "total_degree_weighted"]])
```
