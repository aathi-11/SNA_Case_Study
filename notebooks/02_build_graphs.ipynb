# 02. Graph Construction & Gephi Export

This notebook demonstrates how to construct directed, weighted passing graphs in **NetworkX** and export them to **GraphML** and **Gephi CSV** formats for interactive visualization in Gephi.

```python
import os
from pathlib import Path
import pandas as pd
import networkx as nx
from src.graphs import build_team_graph, export_graphml, export_gephi_csv

# Paths
DATA_DIR = Path("../data")
PROCESSED_DIR = DATA_DIR / "processed"
GEPHI_DIR = DATA_DIR / "gephi"

print("Gephi output directory:", GEPHI_DIR.resolve())
```

## 1. Load Sample Team Match & Construct NetworkX Graph

```python
team_summary = pd.read_parquet(PROCESSED_DIR / "team_match_summary.parquet")
sample_row = team_summary.iloc[0]
match_id = sample_row["match_id"]
team_name = sample_row["team_name"]

print(f"Building passing graph for {team_name} (Match ID: {match_id})")
G = build_team_graph(match_id, team_name)

print(f"Nodes (Players): {G.number_of_nodes()}")
print(f"Edges (Passes): {G.number_of_edges()}")
print("Sample node attribute:", list(G.nodes(data=True))[0])
print("Sample edge attribute:", list(G.edges(data=True))[0])
```

## 2. Export Graph to GraphML & Gephi CSV Formats

```python
safe_team = team_name.replace(" ", "_").lower()

# Export GraphML
gml_path = GEPHI_DIR / f"match_{match_id}_{safe_team}.graphml"
export_graphml(G, gml_path)
print(f"Exported GraphML file: {gml_path}")

# Export Gephi Nodes & Edges CSV
nodes_csv, edges_csv = export_gephi_csv(G, f"match_{match_id}_{safe_team}")
print(f"Exported Gephi Nodes CSV: {nodes_csv}")
print(f"Exported Gephi Edges CSV: {edges_csv}")
```
